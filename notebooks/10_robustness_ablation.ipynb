# 10 — Robustness and ablation

A single tuned FNO result is weak evidence. Change one modeling dimension at a time and record forecast skill, scale-resolved skill, and computational cost.

The purpose is not to find the largest model. It is to test whether the main conclusion survives reasonable design choices.

```python
experiment_grid = {
    "modes": [8, 12, 16, 24],
    "width": [32, 48, 64],
    "lookback_days": [7, 14, 28],
    "horizon_days": [1, 7, 14],
    "spatial_coarsening": [1, 2, 4],
    "random_seed": [11, 42, 73],
}
experiment_grid
```

## Required stress tests

### Fourier modes
If FNO gains disappear when mode count changes slightly, the result is fragile. Record both aggregate persistence skill and frequency-band error skill.

### Season
Report predefined meteorological seasons. Do not discover favorable seasons post hoc and present them as confirmatory evidence.

### Persistence-difficulty regime
Use the threshold estimated from validation in notebook 05. Do not redefine “hard cases” using test quantiles.

### Extreme regional SST
Define high/low anomaly thresholds from training only, then freeze them.

### Resolution
Train at one resolution and evaluate carefully defined regridded cases. Do not call this “zero-shot super-resolution” unless the experiment genuinely supports that statement.

### Fully oceanic spectral subdomain
Repeat the scale attribution on at least one rectangular subdomain with no land mask. If conclusions reverse, mask leakage is a material confounder.

### Seeds
Repeat the finalized learned models over several random seeds.

### Bootstrap block length
Repeat the paired FNO-vs-persistence uncertainty calculation with plausible block lengths (for example 14, 28, and 42 days). The qualitative inference should not hinge on one arbitrary dependence scale.

## Fairness of model comparison

Record approximate tuning budgets for EOF+ridge, CNN/ConvLSTM/U-Net, from-scratch FNO, and official NeuralOperator FNO.

If one architecture receives substantially more search effort or compute, label the comparison accordingly rather than treating it as a clean model-class result.
