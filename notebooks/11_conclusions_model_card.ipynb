The conclusion is about the **specified OISST forecast problem**, not about ocean forecasting in general and not about neural operators universally.

## Questions that must be answered

1. Does FNO beat persistence on the frozen 7-day test forecasts?
2. Does the paired block-bootstrap interval support that improvement?
3. Does FNO beat EOF+ridge?
4. If a local CNN/ConvLSTM/U-Net was trained, does FNO beat it under a comparable tuning budget?
5. At which spatial-frequency bands does FNO gain or lose skill?
6. Does the model gain come from low-frequency smoothing, mid-scale structure, or high-frequency detail?
7. Are gains stable across season and validation-defined difficult persistence cases?
8. Do mask-free spectral subdomains tell the same story?
9. Does the conclusion survive seed, mode-count, horizon, and block-length sensitivity?

## Interpretation templates

Prefer precise statements such as:

> On the frozen 2026 Northeast Atlantic OISST holdout, the FNO reduced aggregate RMSE relative to persistence, but the paired block-bootstrap interval crossed zero; the apparent gain is therefore not robustly distinguished from persistence on this evaluation period.

or:

> The FNO improved aggregate and mid-frequency error relative to persistence and EOF+ridge, while high-frequency error did not improve. The result supports a narrow claim about scale-dependent forecast skill, not a general claim of FNO superiority.

or, if appropriate:

> The FNO did not outperform persistence/EOF+ridge. For this SST-only 7-day task, the added spectral neural-operator complexity was not justified by the evidence.

## Novelty statement

The final write-up must explicitly state that:

- FNO is an established method;
- neural operators have already been applied to regional ocean forecasting;
- Atlantic SST has already appeared in FNO-based spatiotemporal forecasting literature;
- this repository's contribution is an empirical **skill-attribution and model-necessity analysis**, not a first application.

If the completed analysis does not reveal a generalizable new finding, describe the work as a strong portfolio/benchmark study rather than forcing a paper novelty claim.

## Threats to validity to carry into every public result

- OISST is bias-adjusted and optimum-interpolated.
- SST persistence is strong.
- forecast dates are temporally dependent;
- SST-only inputs omit important forcing variables;
- one region/product limits external validity;
- land masks affect Fourier spectra;
- tuning-budget differences can bias model comparisons;
- test-defined exploratory subgroups are not confirmatory evidence.
