# 00 — Problem framing

## Primary question

**When does a compact Fourier Neural Operator add forecast information beyond persistence for Northeast Atlantic SST, and at which spatial scales does that gain occur?**

The supervised forecasting task is

\[
\mathcal G_\theta:(SST_{t-13},\ldots,SST_t)\mapsto SST_{t+7}.
\]

This is **not** presented as a novel ocean-FNO application. FNO-based ocean forecasting already exists, and 2026 work includes Atlantic SST as a real-data FNO application. The narrower goal is diagnostic: determine whether global spectral mixing adds value beyond temporal persistence, low-rank linear dynamics, and local learned spatial filtering.

Read `docs/RESEARCH_POSITIONING.md` before interpreting results.

## Falsifiable hypotheses

### H0 — no useful FNO skill beyond persistence

\[
Skill_{FNO|pers}=1-\frac{RMSE_{FNO}}{RMSE_{pers}}\leq 0.
\]

A sophisticated architecture is not evidence by itself.

### H1 — aggregate skill
The FNO has positive out-of-sample persistence skill on the frozen test period.

### H2 — scale-specific skill
Any aggregate FNO gain should be attributable to identifiable spatial-frequency bands rather than an unexplained scalar RMSE improvement.

### H3 — nonlinear/global value beyond a simple alternative
The FNO should outperform an EOF+ridge baseline that already models global low-rank spatial dynamics. Otherwise the apparent gain may not require nonlinear Fourier operator layers.

### H4 — robustness
The result should not depend entirely on one season, one random seed, or forecast cases that persistence already solves easily.

## Predefined evaluation rules

- Chronological train/validation/test splits only.
- Fit normalizers, climatologies, EOF/PCA bases, and learned baselines using training data only.
- Select hyperparameters using validation only.
- Do not tune frequency bands after seeing test results.
- Define any “hard persistence” test subgroup using a threshold estimated on validation.
- Compare every model on identical target dates and valid cells.
- Use paired forecast-date errors.
- Respect temporal autocorrelation when estimating uncertainty; do not treat daily errors as IID.
- Keep post-hoc subgroup discoveries explicitly labeled exploratory.

## Why FNO is plausible — and why that is not enough

A Fourier layer can exchange information globally through retained spatial modes. That is plausibly useful for a gridded geophysical field with strong large-scale spatial structure.

But three alternative explanations must be ruled out:

1. **persistence:** SST changes slowly;
2. **low-rank dynamics:** a small number of EOF/PCA modes may capture most forecastable variation;
3. **product processing:** OISST is bias-adjusted and optimum-interpolated, so some smoothness is introduced by the data product itself.

The project therefore asks **whether the FNO is necessary**, not merely whether it can fit SST.

## Novelty boundary

Do not claim:

- a new FNO architecture;
- the first neural operator for ocean forecasting;
- the first FNO for Atlantic SST;
- discovery of governing ocean equations;
- universal mesh-independent generalization.

A strong result would instead identify **where and why** FNO skill appears beyond strong null models. A strong negative result is equally acceptable.
